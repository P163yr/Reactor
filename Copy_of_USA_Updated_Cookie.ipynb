```python
import re

def clean_cookie_value(value):
    value = value.strip()

    # Remove outer double quotes
    if len(value) >= 2 and value[0] == '"' and value[-1] == '"':
        value = value[1:-1]

    # Remove outer single quotes
    if len(value) >= 2 and value[0] == "'" and value[-1] == "'":
        value = value[1:-1]

    return value


def parse_trustmrr_cookie_table(raw_cookie_table):
    cookies = []
    seen = set()

    wanted_names = {
        "__Host-authjs.csrf-token",
        "__Secure-authjs.callback-url",
        "__Secure-authjs.session-token",
        "datafast_session_id",
        "datafast_visitor_id",
    }

    for line in raw_cookie_table.splitlines():
        line = line.strip()

        if not line:
            continue

        # Browser cookie table is usually tab-separated.
        if "\t" in line:
            parts = line.split("\t")
        else:
            parts = re.split(r"\s+", line, maxsplit=2)

        if len(parts) < 2:
            continue

        name = parts[0].strip()
        value = clean_cookie_value(parts[1].strip())

        if name not in wanted_names:
            continue

        if name in seen:
            continue

        cookie = {
            "name": name,
            "value": value,
            "url": "https://trustmrr.com/",
            "secure": True,
            "sameSite": "Lax",
        }

        if name.startswith("__Host-authjs") or name.startswith("__Secure-authjs"):
            cookie["httpOnly"] = True
        else:
            cookie["httpOnly"] = False

        cookies.append(cookie)
        seen.add(name)

    if not any(c["name"] == "__Secure-authjs.session-token" for c in cookies):
        raise ValueError(
            "Could not find __Secure-authjs.session-token. "
            "Make sure you pasted logged-in TrustMRR cookies."
        )

    return cookies


TRUSTMRR_COOKIES = parse_trustmrr_cookie_table(RAW_COOKIE_TABLE)

print("Cookies parsed:", len(TRUSTMRR_COOKIES))
for c in TRUSTMRR_COOKIES:
    print("-", c["name"], "=", c["value"][:40] + "...")
```

```python
def extract_slug(item):
    item = item.strip()

    if item.startswith("http"):
        match = re.search(r"/startup/([^/?#]+)", item)
        if not match:
            raise ValueError(f"Could not extract slug from URL: {item}")
        return match.group(1)

    return item


async def scrape_trustmrr_json_files(
    startups,
    out_dir="trustmrr_data",
    fetch_revenue=True,
    fetch_gsc=True,
    fetch_visitors=True,
    headless=True,
    wait_seconds=20,
    sleep_seconds=5,
    max_retries=6,
):
    """
    Saves separate JSON files only:

    - {slug}_revenue_all_daily.json
    - {slug}_gsc_data_all.json
    - {slug}_ga_visitors_all.json

    No combined JSON file.
    No automatic downloads.
    """

    out_path = Path(out_dir)
    out_path.mkdir(exist_ok=True)

    all_results = {}

    async with async_playwright() as p:
        browser = await p.chromium.launch(
            headless=headless,
            args=[
                "--no-sandbox",
                "--disable-dev-shm-usage",
            ],
        )

        context = await browser.new_context(
            viewport={"width": 1400, "height": 900},
            user_agent=(
                "Mozilla/5.0 (Windows NT 10.0; Win64; x64; rv:150.0) "
                "Gecko/20100101 Firefox/150.0"
            ),
        )

        await context.add_cookies(TRUSTMRR_COOKIES)

        # Login check
        check_page = await context.new_page()
        await check_page.goto(
            "https://trustmrr.com/api/auth/session",
            wait_until="domcontentloaded",
            timeout=60000,
        )

        session_text = await check_page.text_content("body")
        await check_page.close()

        if not session_text or '"user"' not in session_text:
            await browser.close()
            raise RuntimeError(
                "Login check failed. Your cookies are expired or incorrect. "
                "Copy fresh logged-in TrustMRR cookies from your browser."
            )

        print("Login check passed.")

        for item in startups:
            slug = extract_slug(item)
            page_url = f"https://trustmrr.com/startup/{slug}?period=all"

            print(f"\n========== {slug} ==========")
            print("Opening:", page_url)

            page = await context.new_page()
            token_holder = {"token": None}

            def on_request(request):
                token = request.headers.get("x-api-token")

                if token and not token_holder["token"]:
                    token_holder["token"] = token
                    print("Captured x-api-token automatically.")

            page.on("request", on_request)

            result = {
                "slug": slug,
                "saved_files": {},
                "errors": {},
            }

            try:
                await page.goto(
                    page_url,
                    wait_until="domcontentloaded",
                    timeout=60000,
                )

                # Wait until page request exposes x-api-token
                for _ in range(wait_seconds * 2):
                    if token_holder["token"]:
                        break
                    await page.wait_for_timeout(500)

                token = token_holder["token"]

                if not token:
                    raise RuntimeError(
                        f"Could not capture x-api-token for {slug}. "
                        "Confirm the TrustMRR startup page loads correctly."
                    )

                async def fetch_json_with_retry(endpoint, label):
                    for attempt in range(1, max_retries + 1):
                        response = await page.evaluate(
                            """
                            async ({ endpoint, token }) => {
                                const res = await fetch(endpoint, {
                                    method: "GET",
                                    credentials: "include",
                                    headers: {
                                        "accept": "*/*",
                                        "x-api-token": token
                                    }
                                });

                                const text = await res.text();

                                let parsed = null;
                                try {
                                    parsed = JSON.parse(text);
                                } catch (e) {
                                    parsed = text;
                                }

                                return {
                                    ok: res.ok,
                                    status: res.status,
                                    data: parsed,
                                    text: text
                                };
                            }
                            """,
                            {
                                "endpoint": endpoint,
                                "token": token,
                            },
                        )

                        if response["ok"]:
                            return response["data"]

                        status = response["status"]
                        data = response["data"]

                        if status == 429:
                            reset_ms = None

                            if isinstance(data, dict):
                                reset_ms = data.get("reset")

                            if reset_ms:
                                now_ms = int(time.time() * 1000)
                                wait_ms = max(reset_ms - now_ms + 3000, 5000)
                                wait_time = wait_ms / 1000
                            else:
                                wait_time = 15 * attempt

                            print(
                                f"{label} hit rate limit. "
                                f"Waiting {wait_time:.1f}s before retry "
                                f"{attempt}/{max_retries}..."
                            )

                            await page.wait_for_timeout(int(wait_time * 1000))
                            continue

                        raise RuntimeError(
                            f"Request failed {status}: {response['text'][:1000]}"
                        )

                    raise RuntimeError(
                        f"{label} failed after {max_retries} retries due to rate limits."
                    )

                # 1. Revenue
                if fetch_revenue:
                    try:
                        endpoint = f"/api/startup/revenue/{slug}?granularity=daily&period=all"
                        print("Fetching revenue:", endpoint)

                        data = await fetch_json_with_retry(endpoint, "Revenue")

                        file_path = out_path / f"{slug}_revenue_all_daily.json"
                        file_path.write_text(
                            json.dumps(data, indent=2),
                            encoding="utf-8",
                        )

                        result["saved_files"]["revenue"] = str(file_path)
                        print("Saved:", file_path)

                    except Exception as e:
                        result["errors"]["revenue"] = str(e)
                        print("Revenue failed:", e)

                    await page.wait_for_timeout(
                        int((sleep_seconds + random.uniform(0.5, 1.5)) * 1000)
                    )

                # 2. GSC / Impressions
                if fetch_gsc:
                    try:
                        endpoint = f"/api/startup/gsc-data/{slug}?period=all"
                        print("Fetching GSC/impressions:", endpoint)

                        data = await fetch_json_with_retry(endpoint, "GSC/impressions")

                        file_path = out_path / f"{slug}_gsc_data_all.json"
                        file_path.write_text(
                            json.dumps(data, indent=2),
                            encoding="utf-8",
                        )

                        result["saved_files"]["gsc_data"] = str(file_path)
                        print("Saved:", file_path)

                    except Exception as e:
                        result["errors"]["gsc_data"] = str(e)
                        print("GSC/impressions failed:", e)

                    await page.wait_for_timeout(
                        int((sleep_seconds + random.uniform(0.5, 1.5)) * 1000)
                    )

                # 3. GA Visitors
                if fetch_visitors:
                    try:
                        endpoint = f"/api/startup/ga-visitors/{slug}?period=all"
                        print("Fetching GA visitors:", endpoint)

                        data = await fetch_json_with_retry(endpoint, "GA visitors")

                        file_path = out_path / f"{slug}_ga_visitors_all.json"
                        file_path.write_text(
                            json.dumps(data, indent=2),
                            encoding="utf-8",
                        )

                        result["saved_files"]["ga_visitors"] = str(file_path)
                        print("Saved:", file_path)

                    except Exception as e:
                        result["errors"]["ga_visitors"] = str(e)
                        print("GA visitors failed:", e)

                    await page.wait_for_timeout(
                        int((sleep_seconds + random.uniform(0.5, 1.5)) * 1000)
                    )

            except Exception as e:
                result["errors"]["page"] = str(e)
                print("Page failed:", e)

            finally:
                all_results[slug] = result
                await page.close()

            # Extra pause between startups
            await asyncio.sleep(sleep_seconds + random.uniform(1.0, 3.0))

        await browser.close()

    return all_results
```
